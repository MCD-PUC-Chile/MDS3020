### 1.1 ¿Qué es NumPy y por qué usarlo?

**NumPy** es la librería fundamental para cómputo científico en Python. Proporciona:
- Un tipo de dato central: el **`ndarray`** (array n-dimensional), compacto y homogéneo.
- **Operaciones vectorizadas** y **broadcasting** para cálculos rápidos sin bucles explícitos.
- Funciones matemáticas de alto desempeño.

> *Idea clave*: con NumPy, trabajas con **bloques** de datos (arrays) en vez de elemento por elemento, lo que suele ser **más rápido** y **más claro** para el análisis numérico.


```python

import numpy as np

print("NumPy importado correctamente.")
print("Versión de NumPy:", np.__version__)
```


<a id="crear"></a>
### 1.2 Crear arrays

Formas típicas de crear arrays:
- A partir de **listas** o **tuplas** con `np.array(...)`.
- Rutas rápidas: `np.arange`, `np.linspace`.
- Inicializadores: `np.zeros`, `np.ones`, `np.full`, `np.eye` (matriz identidad).
- Aleatorios: `np.random.rand`, `np.random.randn`, `np.random.randint`.

> Los arrays son **homogéneos** (todos los elementos comparten el mismo `dtype`).  
> Si quieres forzar un tipo, usa `dtype=...` o `astype(...)`.


```python

# A partir de una lista
x = np.array([[1, 2, 3], [4, 5, 6]])
print("Array x creado a partir de lista:")
print(x)

# Rutas rápidas
a1 = np.arange(0, 10, 2)     # de 0 a 8 de 2 en 2
a2 = np.linspace(0, 1, 5)    # 5 puntos entre 0 y 1 (incluye extremos)
print("\nnp.arange(0, 10, 2):", a1)
print("np.linspace(0, 1, 5):", a2)

# Inicializadores
z = np.zeros((2, 3), dtype=float)
print("\nMatriz de ceros (2x3):\n", z)

o = np.ones((2, 3))
print("Matriz de unos (2x3):\n", o)

f = np.full((2, 3), fill_value=7)
print("Matriz llena con 7 (2x3):\n", f)

e = np.eye(3)  # identidad 3x3
print("Matriz identidad (3x3):\n", e)

# Aleatorios (Opcional: fijar semilla para reproducibilidad)
rng = np.random.default_rng(seed=41)
r1 = rng.random((2, 3))       # U(0,1)
r2 = rng.integers(0, 10, size=(2, 3))  # enteros en [0,10)
print("\nAleatorios uniformes (2x3):\n", r1)
print("Aleatorios enteros (2x3 en [0,10)):\n", r2)
```


<a id="propiedades"></a>
### 1.3 Propiedades importantes (`dtype`, `shape`, `ndim`, `size`, `nbytes`)

Estas propiedades te dicen **cómo está compuesto** tu array:
- `dtype`: tipo de dato de los elementos (por ejemplo `int64`, `float32`).
- `shape`: tupla con el tamaño de cada dimensión (filas, columnas, ...).
- `ndim`: número de dimensiones.
- `size`: cantidad total de elementos.
- `nbytes`: memoria ocupada **solo por los datos** (en bytes).


```python

x = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int64)

print("Array x:\n", x)
print("Tipo de dato (dtype):", x.dtype)
print("Forma (shape):", x.shape)
print("Número de dimensiones (ndim):", x.ndim)
print("Cantidad total de elementos (size):", x.size)
print("Memoria de datos (nbytes):", x.nbytes)

# Convertir de tipo (por ejemplo, a float)
x_float = x.astype(np.float32)
print("\nArray convertido a float32:\n", x_float)
print("Nuevo dtype:", x_float.dtype)
```


<a id="operaciones"></a>
### 1.4 Operaciones básicas y agregaciones

Con arrays puedes operar de forma **vectorizada**:
- Operaciones elemento a elemento: `+`, `-`, `*`, `/`, `**`
- Agregaciones: `np.sum`, `np.mean`, `np.min`, `np.max`, `np.std`
- **Ejes**: usa el parámetro `axis` para operar por filas o columnas.
- **Broadcasting**: operaciones entre arrays de formas compatibles (p.ej. sumar un escalar o un vector a una matriz).


```python

a = np.array([1, 2, 3, 4, 5], dtype=float)
b = np.array([10, 20, 30, 40, 50], dtype=float)

print("a:", a)
print("b:", b)

# Operaciones elemento a elemento
print("\nSuma elemento a elemento (a + b):", a + b)
print("Producto elemento a elemento (a * b):", a * b)
print("Potencia (a ** 2):", a ** 2)

# Agregaciones
M = np.array([[1, 2, 3],
              [4, 5, 6]], dtype=float)
print("\nMatriz M:\n", M)
print("Suma total (np.sum(M)):", np.sum(M))
print("Promedio global (np.mean(M)):", np.mean(M))
print("Mínimo global:", np.min(M), " | Máximo global:", np.max(M))

# Por ejes
print("\nSuma por filas (axis=1):", np.sum(M, axis=1))
print("Suma por columnas (axis=0):", np.sum(M, axis=0))
print("Promedio por columnas (axis=0):", np.mean(M, axis=0))

```

```python
M = np.array([[1, 2, 3],
              [4, 5, 6]], dtype=float)

print("\nMatriz M:\n", M)
# Broadcasting: sumar escalar y sumar vector por filas
print("\nBroadcasting con escalar (M + 10):\n", M + 10)

v = np.array([100, 200, 300], dtype=float)  # se "alinea" con las columnas
print("Broadcasting con vector v=[100,200,300] (M + v):\n", M + v)

# Producto punto vs multiplicación elemento a elemento
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print("\nProducto elemento a elemento (x * y):", x * y)
print("Producto punto (x dot y):", np.dot(x, y))  # 1*4 + 2*5 + 3*6


M = np.array([[1, 2, 3],
              [4, 5, 6]], dtype=float)

transpuesta_M = np.transpose(M)
## Tambien funciona transpuesta_M = M.T


print("\nMatriz Transpuesta M:\n", transpuesta_M)
```

### 1.5 `reshape`, `ravel` y `flatten`

- `reshape` **reorganiza** la forma (shape) sin cambiar los datos. A menudo devuelve una **vista** (comparte memoria).
- `ravel` devuelve un arreglo 1D **aplanado**, intentando ser vista.
- `flatten` devuelve una **copia** 1D.

 **Atención a vistas vs copias**: si dos arrays comparten memoria (vista) y modificas uno, verás el cambio en el otro.


```python

A = np.arange(12)
print("A:", A)
B = A.reshape(3, 4)   # 3x4
print("B = A.reshape(3,4):\n", B)
print("B.shape:", B.shape, "| B.ndim:", B.ndim)

```

```python
# Ravel y flatten
r = B.ravel()
f = B.flatten()
print("\nravel() produce (posible vista) r:", r)
print("flatten() produce (copia) f:", f)
```

```python
# Si cambiamos B, ¿se ve en r y f?
B[0, 0] = 999
print("\nTras modificar B[0,0]=999:")
print("B:\n", B)
print()
print("r (ravel):", r)   # suele reflejar cambio (vista)
print("f (flatten):", f) # copia, no cambia
```

```python


# Si cambiamos r, También cambia B
r[1] = 998
print("\nTras modificar r[1]=998:")
print("r (ravel):", r)
print("B:\n", B)

```

```python

# Si cambiamos f, No cambia B
f[2] = -10
print("\nTras modificar f[1]=997:")
print("r (ravel):", f)
print("B:\n", B)
```


<a id="slicing"></a>
### 1.6 Indexación, slicing y máscaras

- **Slicing** devuelve **vistas** cuando es posible (modificar la subvista afecta el array original).
- **Indexación booleana**: filtra por condición (`arr[arr > 0]`).
- **Indexación avanzada**: seleccionar múltiples índices con listas/arrays.


```python
C = np.arange(1, 10).reshape(3, 3)
print("C:\n", C)

# Slicing (submatriz: filas 0..1, columnas 1..2)
sub = C[0:2, 1:3]
print("\nSubmatriz C[0:2, 1:3]:\n", sub)

# Modificar subvista afecta a C
sub[0, 0] = 777
print("\nTras sub[0,0]=777 -> C se ve afectada:")
print(C)

# Más slicing
fila_1 = C[1, :]      # fila completa
col_2  = C[:, 2]      # columna completa
print("\nFila 1 completa:", fila_1)
print("Columna 2 completa:", col_2)
```

```python

# Máscaras booleanas
mask = C % 2 == 0     # pares
print("\Pares (C % 2 == 0):\n", mask)
print("Elementos pares:", C[mask])

# Indexación avanzada (elegir elementos en posiciones específicas)
idx_rows = np.array([0, 2])
idx_cols = np.array([0, 1])
print("\nElementos C[idx_rows, idx_cols]:", C[idx_rows, idx_cols])
```


<a id="comparacion"></a>
### 1.7 Listas vs Arrays: **rendimiento** y **funcionalidad**

- **Rendimiento**: operaciones vectorizadas con NumPy suelen ser más rápidas que bucles en Python puro.
- **Memoria**: los arrays almacenan datos de forma contigua y homogénea; las listas almacenan referencias a objetos (overhead).
- **Funcionalidad**: broadcasting, operaciones matemáticas avanzadas, álgebra lineal (`@`, `np.linalg.*`).


```python

import timeit

N = 1000000  # 1 millon elementos para ensayar
lista = list(range(N))
arr = np.arange(N, dtype=np.int64)

def suma_lista():
    return sum(lista)

def suma_numpy():
    return np.sum(arr)

t_lista = timeit.timeit(suma_lista, number=3)
t_numpy = timeit.timeit(suma_numpy, number=3)

print("Tiempo sum(lista)  (3 repeticiones):", t_lista, "segundos")
print("Tiempo np.sum(arr) (3 repeticiones):", t_numpy, "segundos")

# Ejemplo de funcionalidad: producto matricial
A = np.arange(9).reshape(3, 3)
B = np.arange(9, 18).reshape(3, 3)
print("\nProducto matricial (A @ B):\n", A @ B)
```


<a id="buenas-practicas"></a>
### 1.8 Errores comunes y buenas prácticas


- **Tipos (`dtype`)**: especifica tipos cuando importe (`float32` vs `float64`) si rendimiento/memoria es crítico.
- **Vectoriza** donde sea posible: evita bucles Python innecesarios.
- **Fija semillas** para reproducibilidad en aleatorios (`np.random.default_rng(seed=...)`).
- **`axis` correcto** en agregaciones (`axis=0` por columnas, `axis=1` por filas).
